## NumPy Arrays

### Types of arrays
- 1D array → vector
- 2D array → matrix
- ND array → tensor

NumPy stores data in homogeneous format (same data type).

### 1D Array

### Array Properties

- shape → structure of array
- ndim → number of dimensions
- size → total elements
- dtype → data type of elements
- reshape → Changing structure without changing data.

```python
arr1 = np.array([1, 2, 3, 4, 5,6])
```

```python
print("1D Array:\n", arr1)
```

```python
print("Shape:", arr1.shape)
```

```python
print("Dimensions:", arr1.ndim)
```

```python
print("Total Elements:", arr1.size)
```

```python
print("Data Type:", arr1.dtype)
```

```python
new_arr = arr1.reshape(2, 3)
print(new_arr)
```

### Vectorized Operations

NumPy allows element-wise operations without loops.
This is called vectorization → very fast computation.

```python
print("Addition:\n", arr1 + 10)
```

```python
print("Multiplication:\n", arr1 * 2)
```

```python
print("Square:\n", arr1 ** 2)
```

```python
print("Division:\n", arr1 / 2)
```

### Statistics in NumPy

- mean → average
- median → middle value
- std → data spread
- max → find maximum
- min → find minimum

```python
print("Mean (Average): ", np.mean(arr1))
```

```python
print("Median (Middle Value): ", np.median(arr1))
```

```python
print("Minimum: ", np.min(arr1))  
```

```python
print("Maximum: ", np.max(arr1)) 
```

```python
print("Standard Deviation: ", np.std(arr1))   
```

### Indexing & Slicing

- Indexing → accessing single value
- Slicing → accessing range of values
- View(Slice) -> Change Original Array
- Copy(Array) -> Doesn't Change Original Array

```python
print("Element at index 0:", arr1[0])
```

```python
print("Element at index 3:", arr1[3])
```

```python
print("Last element:", arr1[-1])
```

```python
print("Slice [1:4]:", arr1[1:4])
```

```python
print("Every 2nd element:", arr1[::2])
```

```python
# VIEW (slicing)
arr1=np.array([1,2,3,4,5])
print("Before changing:", arr1)
slice_arr = arr1[1:4]
slice_arr[0] = 999   # changes original
print("After changing:", arr1)
print("Slice:", slice_arr)
```

```python
# COPY
arr1=np.array([1,2,3,4,5])
print("Before changing:", arr1)
copy_arr = arr1.copy()
copy_arr[0] = 555
print("After changing:", arr1)
print("Copy:", copy_arr)
```
