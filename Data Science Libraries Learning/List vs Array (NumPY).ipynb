# Python List vs NumPy Array Performance

This example compares the execution speed of:

- Normal Python Lists
- NumPy Arrays

The goal is to show why NumPy is faster for numerical computations.


```python
import numpy as np
```

```python
import time 

py_list=list(range(1_000_000))
py_result=[]
start_time=time.time()
for i in py_list:
    py_result.append(i+5)
end_time=time.time()
print("Python List Time: ", end_time - start_time)

np_array=np.arange(1_000_000)
start_time=time.time()
np_result=np_array+5
end_time=time.time()
print("Numpy Array Time: ", end_time - start_time)
```
