# Coin Flip Probability Simulation using NumPy

This example simulates **100,000 coin flips** using NumPy random functions.

The program estimates:
- Probability of Heads
- Probability of Tails

A coin has:
- `1` → Heads
- `0` → Tails

```python
!pip install numpy
```

```python
# simulate 100k coin flips
rng=np.random.default_rng()
flips=rng.integers(0,2,size=100000)
prob_heads=np.mean(flips)
prob_tails=1-prob_heads
print("Estimated probability of heads: ", prob_heads)
print("Estimated probability of tails: ", prob_tails)
print("Total: ", prob_heads+prob_tails)
```

#### Seed to stablility

```python
# simulate 100k coin flips with fixed seed
rng=np.random.default_rng(seed=42)
flips=rng.integers(0,2,size=100000)
prob_heads=np.mean(flips)
prob_tails=1-prob_heads
print("Estimated probability of heads: ", prob_heads)
print("Estimated probability of tails: ", prob_tails)
print("Total: ", prob_heads+prob_tails)
```
