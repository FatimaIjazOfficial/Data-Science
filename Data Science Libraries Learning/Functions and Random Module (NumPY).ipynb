```python
import numpy as np
```

# NumPy Functions

- zeros() → array of 0s
- ones() → array of 1s
- full() → array filled with value
- arange() → range with step
- linspace() → equal spaced values
- eye() → identity matrix
- diag() → diagonal matrix / diagonal extract
- dtype → data type check
- astype() → change data type

```python
print(np.zeros((2,3)))
```

```python
print(np.ones((2,3)))
```

```python
print(np.full((2,3),7))  #full(size,number)
```

```python
print(np.arange(0,10,2)) #arange(start,end-1,steps)
```

```python
print(np.linspace(0,1,5))  #linspace(start,end,complete in steps)
```

```python
print(np.eye(3))         #identity matrix
```

```python
print(np.diag([1,2,3]))
```

```python
print(np.array([1,2,3],dtype=float))
```

```python
print(np.array([1.5,2.7,3.9]).astype(int))
```

# Random module

```python
print(np.random.rand(3))      # random values (0–1)
```

```python
print(np.random.randint(1,10)) # random integer
```

```python
print(np.random.randn(3,3))   # normal distribution
```
