## NumPy Arrays

### Types of arrays
- 1D array → vector
- 2D array → matrix
- ND array → tensor

NumPy stores data in homogeneous format (same data type).

### 2D Array

### Array Properties

- shape → structure of array
- ndim → number of dimensions
- size → total elements
- dtype → data type of elements
- reshape → Changing structure without changing data.

```python
arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])
```

```python
print("2D Array:\n", arr2)
```

```python
print("Shape (rows, columns):", arr2.shape)
```

```python
print("Dimensions:", arr2.ndim)
```

```python
print("Total Elements:", arr2.size)
```

```python
print("Data Type:", arr2.dtype)
```

```python
new_arr = arr2.reshape(3,2)
print(new_arr)
```

### Vectorized Operations

NumPy allows element-wise operations without loops.
This is called vectorization → very fast computation.

```python
print("Addition:\n", arr2 + 10)
```

```python
print("Multiplication:\n", arr2 * 2)
```

```python
print("Square:\n", arr2 ** 2)
```

```python
print("Division:\n", arr2 / 2)
```

### Statistics in NumPy

- mean → average
- median → middle value
- std → data spread
- max → find maximum
- min → find minimum

```python
print("Mean (Average): ", np.mean(arr2))
```

```python
print("Median (Middle Value): ", np.median(arr2))
```

```python
print("Minimum: ", np.min(arr2))  
```

```python
print("Maximum: ", np.max(arr2)) 
```

```python
print("Standard Deviation: ", np.std(arr2))   
```

### Indexing & Slicing

- Indexing → accessing single value
- Slicing → accessing range of values
- View(Slice) -> Change Original Array
- Copy(Array) -> Doesn't Change Original Array

```python
print("Element [0,1] (row0,col1):", arr2[0, 1])
```

```python
print("Element [1,2] (row1,col2):", arr2[1, 2])
```

```python
print("Negative Index [-1,-2]:", arr2[-1, -2])   # last row, 2nd last col
```

```python
print("Negative Index [-2,-1]:", arr2[-2, -1])   # 1st row, last col
```

```python
print("Slice rows 0-1, cols 1-2:\n", arr2[0:2, 1:3])
```

```python
print("Every 2nd row & col:\n", arr2[::2, ::2])
```

```python
# VIEW (slicing)
arr2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print("Before changing:\n", arr2)
slice_arr = arr2[0, :2]
slice_arr[0] = 999   # changes original
print("After changing:\n", arr2)
print("Slice:", slice_arr)
```

```python
# COPY
arr2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print("Before changing:\n", arr2)
copy_arr = arr2.copy()
copy_arr[0,0] = 555
print("After changing:\n", arr2)
print("Copy:\n", copy_arr)
```
